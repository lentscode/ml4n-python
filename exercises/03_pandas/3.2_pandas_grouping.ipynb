## $(\star)$ Exercise 3 - predict the class of a sample by computing the distance from the mean of each class
- Create a function to compute the distance of a sample from the mean of each class 
- Then create a new column in the dataset called "predicted_label" and assign the predicted class to each sample
- Then compute the accuracy of the prediction

**Note:** you must not iterate over the dataframe


```python
import numpy as np

def distance(df: pd.DataFrame):
    means = df.groupby("label").mean(numeric_only=True)
    distances = {}
    for c in df["label"].unique():
        for key in df.columns:
            if key == "label" or key == "predicted_label": continue
            if c not in distances.keys():
                distances[c] = (df[key] - means.loc[c][key]) ** 2
            else:
                distances[c] += (df[key] - means.loc[c][key]) ** 2
        distances[c] = np.sqrt(distances[c])

    labels = []
    for i in range(df.shape[0]):
        label, val = None, 0
        for c in df["label"].unique():
            if distances[c][i] < val or label is None:
                label, val = c, distances[c][i]
        labels.append(label)

    df["predicted_label"] = labels
    return 1 - (df[df["label"] != df["predicted_label"]].shape[0] / df.shape[0])
    

accuracy = distance(df)
accuracy
```
