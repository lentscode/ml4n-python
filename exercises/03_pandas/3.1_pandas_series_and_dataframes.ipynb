## Exercise 1 - Create and access a Pandas Series
Create a pandas series with:
 - the following values:  [1, 3, 2, 7, 5, 4, 9] 
 - indexed with the following fruit names: ["apple", "orange", "cherry", "pear", "pineapple", "mango", "banana"]

Then print:
 - the first three elements of the series
 - all the values associated to a "pear" or a "mango"
 - all the indexes associated to a value greater than 5

````python
s = pd.Series([1, 3, 2, 7, 5, 4, 9], index=["apple", "orange", "cherry", "pear", "pineapple", "mango", "banana"])

s.iloc[:3]
````

````python
s.loc[["pear", "mango"]]
````

````python
s.loc[s > 5].index
````

Given the following column names
['height','width','intensity','weight']

Compute the following:
- Create a pandas dataframe with the given data and column names
- Add a new composite feature, 'area' = 'width' * 'height'
- Retrieve the elements associated to an odd index
- Retrieve the 'height' and 'weight' of the samples with area > 20

````python
df = pd.DataFrame([[5.1, 3.5, 1, 0.2],
[4.3, 3. , 1, 0.1],
[5. , 0. , 1, 0.4],
[5.1, 3.4, 2, 0.2],
[7.0, 3.2, 1, 0.2],
[6.9, 3.1, 3, 1.5],
[6.7, 3.1, 1, 2. ],
[6. , 2.9, 2, 1.5],
[6.1, 3. , 2, 1.4],
[6.5, 3. , 3, 2.2],
[7.7, 3.8, 3, 2.2],
[7.4, 2.8, 1, 1.9],
[6.8, 3.2, 1, 2.3]], columns=['height','width','intensity','weight'])

df
````

````python
df["area"] = df["width"] * df["height"]
df
````

````python
df.iloc[0:len(df):2]
````

````python
df.loc[df["area"] > 20][["height", "weight"]]
````
