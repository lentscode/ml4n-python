Given the lists:

## Exercise 3 creates a ndarray with values without using lists
Create:
1) an ndarray 3 x 3 of $1$s
2) an ndarray 2 x 2 of $0$s
3) an ndarray 1 x 2 of $9$s
4) an ndarray of shape () with value 39
5) an ndarray of shape (2,5) with values randomly distributed between 0 and 1
6) an ndarray with 100 (equispaced) elements between 0 and 5

```python
ones = np.ones((3, 3))
print(ones)

zeros = np.zeros((2, 2))
print(zeros)

nines = np.full((1, 2), 9)
print(nines)

empty = np.full((), 39)
print(empty)

random = np.random.random((2, 5))
print(random)

spaced = np.linspace(0, 5, 100)
print(spaced)
```
